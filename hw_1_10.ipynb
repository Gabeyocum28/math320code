## Problem 1.64

A sequence $(x_i)_{i=0}^{n}$ is _unimodal_ if it consists of an increasing sequence followed by a decreasing sequence; that is, there is some $k \in \{0, \ldots, n\}$ such that $x_{i-1} < x_{i}$ when $0 < i \leq k$, and $x_{i+1} < x_{i}$ when $k \leq i < n$.

### Part (i)

Give an algorithm with temporal complexity $O(\log n)$ that finds the maximal element $x_m$ of a unimodal sequence. (An informal description suffices here)


Look at the middle element and the one right after it. If the second one is bigger, the list is still going up there, so the max is somewhere to the right and we throw away the left half. If not, the list is going down there, so the max is at the middle or to the left, and we throw away the right half. Repeat on the half we kept until only one element is left. That element is the max. Since we cut the list in half every time, this takes O(log n) steps.

### Part (ii)

Code up your algorithm inside the function `max_unimodal()`, and explain the details in the comments.

The function should accept a list `L` of length $n$ containing a unimodal sequence of floating-point numbers, and should
return the value of the maximal element in `L` in $O(\log n)$ time.

```python
def max_unimodal(L: list[float]) -> float:
    """Given a unimodal sequence of floating-point numbers,
    returns the maximal element in O(log(n)) time.

    Args:
        L (list[float]): A list containing a unimodal sequence of floating-point numbers

    Returns:
        float: The maximal element of the unimodal sequence
    """
    # left and right are the bounds of the slice that could still contain the peak.
    # The peak index always satisfies left <= peak <= right.
    left = 0
    right = len(L) - 1

    # Each pass shrinks the slice to (at most) half its size.
    while left < right:
        mid = (left + right) // 2

        if L[mid] < L[mid + 1]:
            # Increasing at mid, so we are on the left side of the peak.
            # The peak is strictly to the right of mid.
            left = mid + 1
        else:
            # Decreasing at mid, so we are on the right side of the peak
            # (or mid is the peak itself). Drop everything after mid.
            right = mid

    # The slice has shrunk to a single index, which must be the peak.
    return L[left]
```

### Part (iii)

Prove the $O(\log n)$ bound on the temporal complexity in the following cell (use LaTeX commands inside of dollar signs \$ \$ to write mathematical notation).

Each time through the loop we only do a constant amount of work: one comparison and one index update. So the running time is just the number of loops times a constant.

Every loop throws away half of the remaining list. Starting with n elements, we can only cut in half about log_2 n times before we are down to one element and the loop stops.

So there are at most about log_2 loops, each taking constant time, which gives O(log n) total.
